```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

filename = "/mnt/lochy/ASPECT_DATA/Collision0/collision_test/S20RTS/regional_velocity_files/chunk_3d_south.0.txt"

data = np.loadtxt(filename, comments="#")

# First two columns
r = data[:, 0]
longitude = data[:, 1]  # radians

# Assign constant latitude
latitude = np.deg2rad(-55.0)  # example: 30 degrees north

# Spherical -> Cartesian
x = r * np.cos(latitude) * np.cos(longitude)
y = r * np.cos(latitude) * np.sin(longitude)
z = r * np.sin(latitude)

# Convert m -> km
x /= 1e3
y /= 1e3
z /= 1e3

# Plot
fig = plt.figure(figsize=(9, 8))
ax = fig.add_subplot(111, projection="3d")

# Connect points along each constant-longitude radial profile
for lon_value in np.unique(longitude):
    mask = np.isclose(longitude, lon_value)

    ax.scatter(
        x[mask],
        y[mask],
        z[mask],
        s=9,
        c="b"
    )

ax.set_xlabel("x (km)")
ax.set_ylabel("y (km)")
ax.set_zlabel("z (km)")

ax.set_box_aspect([
    np.ptp(x),
    np.ptp(y),
    np.ptp(z),
])

plt.tight_layout()
plt.show()
```
